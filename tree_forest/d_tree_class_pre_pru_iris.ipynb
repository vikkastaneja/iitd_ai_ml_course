```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from matplotlib import pyplot as plt

# Function for post pruning for decision tree

from sklearn.datasets import load_iris

# Independent and dependent variables
iris = load_iris()
X = iris.data
y = iris.target
```

```python
df = pd.DataFrame(X, columns=iris.feature_names)
```

```python
df.isnull().sum()
```

```python


# Splitting the data into training and testing data
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

# Decision tree classifier - pre pruning - tune hyper parameters at the time of training
# In order to tune the hyper parameters, we need to know the hyper parameters of the model
# We can use the get_params() method to get the hyper parameters of the model
# We can use the set_params() method to set the hyper parameters of the model
parameters = {
    'criterion': ['gini', 'entropy', 'log_loss'],
    'splitter': ['best', 'random'],
    'max_depth': [1, 2, 3, 4, 5],
    'max_features': ['auto', 'sqrt', 'log2'],
}

# Use GridSearchCV to find the best hyper parameters
from sklearn.model_selection import GridSearchCV

clf = DecisionTreeClassifier()

cv = GridSearchCV(estimator=clf, param_grid=parameters, cv=5)

cv.fit(X_train, y_train)

print('best params: ', cv.best_params_)
print('best score: ', cv.best_score_)

y_pred = cv.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report
score = accuracy_score(y_test, y_pred)
print('Accuracy: ', score)

print(classification_report(y_test, y_pred))
```
